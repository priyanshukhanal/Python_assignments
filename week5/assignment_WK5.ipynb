```python
from abc import ABC, abstractmethod


# Abstract Base Class
class Vehicle(ABC):

    def __init__(self, vehicle_id, model, daily_rate):
        self.vehicle_id = vehicle_id
        self.model = model
        self.__daily_rate = daily_rate

    # Getter method
    def get_daily_rate(self):
        return self.__daily_rate

    # Setter method with validation
    def set_daily_rate(self, daily_rate):
        if daily_rate <= 0:
            raise ValueError("Daily rate must be greater than zero.")

        self.__daily_rate = daily_rate

    # Abstract method
    @abstractmethod
    def calculate_rental_cost(self, days):
        pass

    # Operator overloading for sorting
    def __lt__(self, other):
        return self.get_daily_rate() < other.get_daily_rate()

```

```python
# Passenger Car Class
class PassengerCar(Vehicle):

    def __init__(self, vehicle_id, model, daily_rate, num_seats):
        super().__init__(vehicle_id, model, daily_rate)
        self.num_seats = num_seats

    # Method overriding
    def calculate_rental_cost(self, days):
        return (self.get_daily_rate() * days) + (15.0 * self.num_seats)

    # String representation
    def __str__(self):
        return (
            f"Passenger Car | ID: {self.vehicle_id} | "
            f"Model: {self.model} | Seats: {self.num_seats} | "
            f"Daily Rate: ${self.get_daily_rate():.2f}"
        )
```

```python
# Cargo Truck Class
class CargoTruck(Vehicle):

    def __init__(self, vehicle_id, model, daily_rate, cargo_capacity_tons):
        super().__init__(vehicle_id, model, daily_rate)
        self.cargo_capacity_tons = cargo_capacity_tons

    # Method overriding
    def calculate_rental_cost(self, days):
        return (
            self.get_daily_rate() * days
            + (50.0 * self.cargo_capacity_tons)
        )

    # String representation
    def __str__(self):
        return (
            f"Cargo Truck | ID: {self.vehicle_id} | "
            f"Model: {self.model} | "
            f"Capacity: {self.cargo_capacity_tons} tons | "
            f"Daily Rate: ${self.get_daily_rate():.2f}"
        )
```

```python
# Polymorphic Fleet Manifest Function
def generate_fleet_manifest(fleet_list, days):

    print("\n========== LOGITRANS FLEET MANIFEST ==========")
    print(f"Rental Duration: {days} days")
    print("----------------------------------------------")

    total_cost = 0

    for vehicle in fleet_list:
        print(vehicle)
        rental_cost = vehicle.calculate_rental_cost(days)

        print(f"Rental Cost: ${rental_cost:.2f}")
        print("----------------------------------------------")

        total_cost += rental_cost

    print(f"Total Fleet Rental Cost: ${total_cost:.2f}")
    print("==============================================")
```

```python
# Main Program
if __name__ == "__main__":

    # Create vehicle objects
    car1 = PassengerCar("CAR101", "Toyota Corolla", 50.0, 5)
    car2 = PassengerCar("CAR102", "Honda Civic", 60.0, 4)

    truck1 = CargoTruck("TRK201", "Volvo FH", 120.0, 10.5)
    truck2 = CargoTruck("TRK202", "Mercedes Actros", 150.0, 15.0)

    # Store all vehicles in one list
    fleet = [car1, car2, truck1, truck2]

    # Generate fleet manifest
    generate_fleet_manifest(fleet, 3)

    # Display vehicles sorted by daily rental rate
    print("\nVehicles Sorted by Daily Rental Rate:")

    for vehicle in sorted(fleet):
        print(vehicle)

    # Test getter and setter
    print("\nUpdating CAR101 Daily Rate:")

    print("Original Rate:", car1.get_daily_rate())

    car1.set_daily_rate(55.0)

    print("Updated Rate:", car1.get_daily_rate())

    # Test invalid daily rate
    print("\nTesting Invalid Daily Rate:")

    try:
        car1.set_daily_rate(-10.0)
    except ValueError as error:
        print("Error:", error)
```
