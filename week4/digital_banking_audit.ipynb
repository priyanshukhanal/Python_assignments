```python
# ---------- Task 1: functional tools ----------
 
def clean_transactions(amounts, min_threshold=10.0):
    """Keep only positive amounts that are at least min_threshold."""
    return list(filter(lambda amt: amt > 0 and amt >= min_threshold, amounts))
 
 
def apply_service_fee(amounts, fee_rate=0.02):
    """Deduct a percentage service fee from every amount."""
    return list(map(lambda amt: round(amt * (1 - fee_rate), 2), amounts))
```

```python
# ---------- Task 2: recursive search engine ----------
 
def calculate_nested_total(data):
    """Sum a number or an arbitrarily nested list of numbers (no loops)."""
    # Base case 1: empty list
    if isinstance(data, list) and len(data) == 0:
        return 0.0
    # Base case 2: a single number
    if isinstance(data, (int, float)):
        return data
    # Recursive step: split into head and tail
    head = data[0]
    tail = data[1:]
    return calculate_nested_total(head) + calculate_nested_total(tail)
```

```python
# ---------- Test run ----------
 
if __name__ == "__main__":
    print("=== Task 1: Cleaning and Fees ===")
    raw = [250.0, -40.0, 5.5, 0, 10.0, 99.99, 9.99, 1200.0]
    print("Raw batch        :", raw)
 
    cleaned = clean_transactions(raw)
    print("Cleaned (min 10) :", cleaned)
 
    cleaned_50 = clean_transactions(raw, min_threshold=100.0)
    print("Cleaned (min 100):", cleaned_50)
 
    after_fee = apply_service_fee(cleaned)
    print("After 2% fee     :", after_fee)
 
    after_fee_5 = apply_service_fee(cleaned, fee_rate=0.05)
    print("After 5% fee     :", after_fee_5)
 
    print("\n=== Task 2: Nested Totals ===")
    tests = [
        [100, [50, 25], [[10, 5], 40]],
        [],
        75.5,
        [[[[20]]], 30, [[], [5, [5]]]],
        [1.5, [2.5, [3.5, [4.5]]]],
    ]
    for t in tests:
        print(f"{str(t):<35} -> {calculate_nested_total(t)}")
 
    print("\n=== Full audit pipeline ===")
    batch = [[500, -20, [150, 8]], [300, [12, 3]], 45]
    print("Nested batch:", batch)
    print("Total transferred (nested):", calculate_nested_total(batch))
```
