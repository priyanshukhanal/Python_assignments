```python
input_file = open("raw_logs.txt", "r")
master_file = open("decrypted_master.txt", "w")
alert_file = open("security_alerts.txt", "w")

for line in input_file:

    decrypted_line = ""

    for char in line:

        if char >= 'A' and char <= 'Z':
            decrypted_line = decrypted_line + chr((ord(char) - ord('A') - 3) % 26 + ord('A'))

        elif char >= 'a' and char <= 'z':
            decrypted_line = decrypted_line + chr((ord(char) - ord('a') - 3) % 26 + ord('a'))

        else:
            decrypted_line = decrypted_line + char

    decrypted_line = decrypted_line.strip()

    master_file.write(decrypted_line + "\n")

    if "BREACH" in decrypted_line.upper():
        alert_file.write(decrypted_line + "\n")

input_file.close()
master_file.close()
alert_file.close()

print("Decryption completed successfully")
```
