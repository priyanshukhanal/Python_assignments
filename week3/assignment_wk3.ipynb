```python
# Inventory
inventory = {
    "Widget": 10,
    "Gadget": 5,
    "Sensor": 0,
    "Cable": 15
}

# Order queue
orders = [
    ["Widget", 3],
    ["Sensor", 2],
    ["Gadget", 7],
    ["Cable", 5],
    ["Widget", 10],
    ["Unknown", 2]
]

# Unfulfilled orders
unfulfilled_items = []

# Fully fulfilled order count
fully_fulfilled_count = 0
```

```python
# Process orders
for order in orders:
    item = order[0]
    requested = order[1]

    stock = inventory.get(item)

    if stock is None:
        print(f"ALERT: {item} is not available.")
        unfulfilled_items.append([item, requested])

    elif stock == 0:
        print(f"OUT OF STOCK: {item}")
        unfulfilled_items.append([item, requested])

    elif stock >= requested:
        inventory[item] = stock - requested
        fully_fulfilled += 1
        print(f"SUCCESS: {requested} {item}(s) fully fulfilled.")

    else:
        remaining = requested - stock
        print(f"PARTIAL: {stock} {item}(s) fulfilled. "
              f"{remaining} unfulfilled.")

        inventory[item] = 0
        unfulfilled_items.append([item, remaining])
```

```python
# Analytics & Summary Output
print("\n=== End-of-Batch Summary ===\n")

# 1. Final updated inventory dictionary
print("Final Inventory:")
for item, stock in inventory.items():
    print(f"  {item}: {stock} units")

# 2. Total count of fully fulfilled orders
print(f"\nFully Fulfilled Orders: {fully_fulfilled_count}")

# 3. List of items/quantities that could not be fully supplied
print("\nUnfulfilled Items:")
if len(unfulfilled_items) == 0:
    print("  None")
else:
    for item in unfulfilled_items:
        print(f"  {item[0]}: {item[1]} units")
```
